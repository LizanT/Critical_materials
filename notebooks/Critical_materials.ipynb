```python
mask_gt = df['RESERVES_2024'].astype(str).str.contains('>', na=False) # Delete comma and greather than.
print(df.loc[mask_gt, ['COMMODITY', 'COUNTRY', 'RESERVES_2024']])
```

```python
df['RESERVES_2024_CLEAN'] = ( #remove comma and greather than.
    df['RESERVES_2024']
    .astype(str)
    .str.replace('>', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)
df['RESERVES_2024_CLEAN'] = pd.to_numeric(df['RESERVES_2024_CLEAN'], errors='coerce')
df['RESERVES_2024_CLEAN'].describe()
```

```python
# Separate rows of world total/Other Countries.
mask_world = df['COUNTRY'].str.contains('World total', case=False, na=False)
mask_other = df['COUNTRY'].str.contains('Other Countries', case=False, na=False)
df_world_totals = df[mask_world].copy()
df_by_country = df[~mask_world & ~mask_other].copy()
```

```python
# Country names cleaned.

country_exceptions_kinshasa = 'Democratic Republic of the Congo'

def clean_country(country):
    if 'Kinshasa' in country:
        return country_exceptions_kinshasa
    if 'Brazzaville' in country:
        return 'Republic of the Congo'
    country = country.replace('\xa0', ' ')
    country = re.sub(r'\s*\(.*?\)', '', country)
    return country.strip()

df_by_country['COUNTRY_CLEAN'] = df_by_country['COUNTRY'].apply(clean_country)
df_by_country = df_by_country[df_by_country['COUNTRY_CLEAN'] != 'United States and Canada']
```

```python
# Separate production and reserves.
df_prod_source = df_by_country[df_by_country['PROD_EST_ 2024'].notna()].copy()
df_reserves_source = df_by_country[df_by_country['RESERVES_2024_CLEAN'].notna()].copy()
```

```python
# exclude problematic production lines (Copper refinería, Titanium dioxide).
df_prod_source = df_prod_source[
    ~((df_prod_source['COMMODITY'] == 'Copper') & (df_prod_source['TYPE'] == 'Refinery production, copper'))
]
df_prod_source = df_prod_source[df_prod_source['COMMODITY'] != 'Titanium & titanium dioxide']

print(f"Filas por país: {len(df_by_country)}")
print(f"Filas con producción: {len(df_prod_source)}")
print(f"Filas con reservas: {len(df_reserves_source)}")
```

```python
check = df_prod_source.groupby(['COMMODITY', 'COUNTRY_CLEAN']).size()
print(check[check > 1])
```

```python
# Add Platinum-Group metals, Silicon y Titanium Mineral Concentrates by country/year.
df_prod_grouped = (
    df_prod_source
    .groupby(['COMMODITY', 'COUNTRY_CLEAN'], as_index=False)
    .agg({'PROD_EST_ 2024': 'sum'})
)

print(df_prod_grouped.shape)
df_prod_grouped.head()
```

```python
production_rows = []

for _, row in df_prod_grouped.iterrows():
    commodity = row['COMMODITY']
    if commodity not in csv_to_criticals:
        continue  # If is not as our 55 critical material(ej. Bauxite, Cement...)
    
    criticals = csv_to_criticals[commodity]
    is_agg = len(criticals) > 1  # True for Rare earths and Platinum-Group metals.
    
    for mineral_name in criticals:
        production_rows.append({
            'mineral_name': mineral_name,
            'country_name': row['COUNTRY_CLEAN'],
            'year': 2024,
            'tonnes': row['PROD_EST_ 2024'],
            'is_aggregated': is_agg,
            'source': 'USGS MCS 2025'
        })

df_production_final = pd.DataFrame(production_rows)
print(df_production_final.shape)
df_production_final.head(10)
```
